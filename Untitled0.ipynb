```python
%cd /content/model_mayhem
from pathlib import Path
import shutil, zipfile
from google.colab import files
from infer import run_inference
from src.constants import IMAGE_EXTENSIONS
from validate_submission import validate_submission_csv

HIDDEN_URL = ""  # Optional: paste direct HTTP URL if organizers gave a link

test_dir = Path("data/hidden_test")
shutil.rmtree(test_dir, ignore_errors=True)
shutil.rmtree(Path("vis_samples"), ignore_errors=True)
test_dir.mkdir(parents=True, exist_ok=True)

if HIDDEN_URL.strip():
    !curl -L "{HIDDEN_URL.strip()}" -o hidden_test.zip
    zpath = Path("hidden_test.zip")
else:
    print("Upload the released test .zip (or test image files):")
    up = files.upload()
    names = list(up.keys())
    zips = [n for n in names if n.lower().endswith(".zip")]
    if zips:
        zpath = Path(zips[0])
    else:
        zpath = None
        for n in names:
            shutil.move(n, test_dir / Path(n).name)

if zpath is not None:
    with zipfile.ZipFile(zpath, "r") as zf:
        zf.extractall(test_dir)

test_imgs = sorted(p for p in test_dir.rglob("*") if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS)
assert test_imgs, f"No images found inside {test_dir}"
use_rot = len(test_imgs) <= 500
print(f"Found {len(test_imgs)} test images | rot_tta={use_rot}")

run_inference(
    source=test_dir,
    weights=[Path("weights/best.pt")],
    output_csv=Path("predictions.csv"),
    imgsz=1024,
    conf=0.01,
    iou=0.60,
    tta=(not use_rot),
    rot_tta=use_rot,
    tile_threshold=0,
    recursive=True,
    save_vis=Path("vis_samples"),
    validate=True,
)

ok, errs, stats = validate_submission_csv(Path("predictions.csv"), images_dir=test_dir)
print("Submission CSV Audit:", stats)
assert ok, errs

bundle = Path("ModelMayhem_Final_Submission.zip")
with zipfile.ZipFile(bundle, "w", zipfile.ZIP_DEFLATED) as zf:
    for p in [Path("predictions.csv"), Path("weights/best.pt")]:
        if p.exists():
            zf.write(p, p.as_posix())
    for p in sorted(Path("vis_samples").glob("*.jpg"))[:12]:
        zf.write(p, p.as_posix())
files.download(str(bundle))
```
